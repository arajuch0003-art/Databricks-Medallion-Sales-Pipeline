```python
# Bronze table nundi read cheyadam
df_bronze = spark.table("sales_bronze")

# Transformation: 10% Discount column add cheyadam
from pyspark.sql.functions import col

df_silver = df_bronze.withColumn("discounted_amount", col("amount") * 0.9)

# Silver Delta Table ga save cheyadam
df_silver.write.format("delta").mode("overwrite").saveAsTable("sales_silver")
print("🥈 Sales Silver Table Created Successfully!")
```

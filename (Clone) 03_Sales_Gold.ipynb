```python
# Silver table nundi read cheyadam
df_silver = spark.table("sales_silver")

# Aggregation: Total Sales Amount per Date
from pyspark.sql.functions import sum

df_gold = df_silver.groupBy("sale_date").agg(
    sum("discounted_amount").alias("total_sales_revenue")
)

# Gold Delta Table ga save cheyadam
df_gold.write.format("delta").mode("overwrite").saveAsTable("sales_gold")
print("🥇 Sales Gold Table Created Successfully!")
```
